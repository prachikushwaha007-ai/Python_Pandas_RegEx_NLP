```python
# Run SpaCy with GPU is more efficient.
import spacy
```

```python
with open("/content/drive/MyDrive/CHAPTER I.txt", "r") as f:
  text = f.read()
print(text)
```

```python
# to cleanup the line breaks.
text = text.replace("\n\n", " ").replace("\n", " ")
```

```python
chapters = text.split("CHAPTER ")[1:]
```

```python
chapter1 = chapters[0]
print(chapter1)
```

```python
# downloading large module.
!python -m spacy download en_core_web_lg

nlp = spacy.load("en_core_web_lg")
```

```python
#nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp(chapter1)

sentences = list(doc.sents)
print(sentences[1])
```

```python
# Named Entity Recognition
# make a series of entities using ents object.

sentence = sentences[1]

ents = list(doc.ents)

people = []
for ent in ents:
  if ent.label_ == "PERSON":
    people.append(ent)

print(people)

print(ents[0].label)

print(ents[0].label_)

print(ents[0].text)
```
