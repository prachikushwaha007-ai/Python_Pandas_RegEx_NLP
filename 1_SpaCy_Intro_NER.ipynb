```python

import spacy
```

```python
# To Openup Text File.

with open("/content/drive/MyDrive/CHAPTER I.txt", "r") as f:
  text = f.read()
print(text)
```

```python
# To cleanup the line breaks.
text = text.replace("\n\n", " ").replace("\n", " ")
```

```python
chapters = text.split("CHAPTER ")[1:]
```

```python
chapter1 = chapters[0]
print(chapter1)
```

```python
# downloading large module.
!python -m spacy download en_core_web_lg

nlp = spacy.load("en_core_web_lg")
```

```python
#nlp = spacy.load("en_core_web_sm")
```

```python
# Make a Document of chapter.

doc = nlp(chapter1)

# To Make a List of Sentence.

sentences = list(doc.sents)
print(sentences[1])
```

```python
# Named Entity Recognition
# make a series of entities using ents object.

sentence = sentences[1]

ents = list(doc.ents)

people = []
for ent in ents:
  if ent.label_ == "PERSON":
    people.append(ent)

print(people)

print(ents[0].label)

print(ents[0].label_)

print(ents[0].text)
```

```python
# To Find part of Speech.

for token in sentence:
   print(token.text, token.pos_)
```

```python
# To Extract All Nouns from Sentence.

nouns = []

for token in sentence:
  if token.pos_ == "NOUN":
     nouns.append(token)

print(nouns)
```

```python
# To Extract All Adjectives from Sentence.

adj = []

for token in sentence:
  if token.pos_ == "ADJ":
    adj.append(token)

print(adj)
```

```python
# To Identify Groupings of Nouns across the Entire Chapter.

print(list(doc.noun_chunks))
```

```python
# to Find Specific Noun.

chunks = (list(doc.noun_chunks))

for chunk in chunks:
  if 'watch' in str(chunk):
     print(chunk)
```

```python
# to Extract Places.

places = []

for ent in doc.ents:
   if ent.label_ in ['GPE','LOC']:
      places.append(ent.text, ent.label_)
print(places)
```

```python
# Lemmatization for finding the roots for quick searching.

for word in sentence:
  if word.pos_ == 'NOUN':
    print(word, word.lemma_)
```

```python
# matcher.
from spacy.matcher import Matcher
```

```python
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN", "OP": "+"}]
matcher.add("PROPER_NOUNS", [pattern], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN", "OP": "+"}, {"POS": "VERB", "OP": "+"}]
matcher.add("PROPER_NOUNS", [pattern], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
matcher = Matcher(nlp.vocab)
pattern = [{"ORTH": "“"},
           {"IS_ALPHA": True, "OP": "+"},
           {"IS_PUNCT": True, "OP": "?"},
           {"ORTH": "”"}
          ]
matcher.add("PROPN", [pattern], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
  print(match, doc[match[1]:match[2]])
```
