```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp("Britain is a place. Mary is a doctor. Mary is british.")
```

```python
for ent in doc.ents:
   print(f"Entity: {ent.text} | Label: {ent.label_}")
```

```python
# to remove gpe and set as loc, make custom components.

from spacy.language import Language
```

```python
@Language.component("remove_norp")
def remove_norp(doc):
   original_ents = list(doc.ents)
   for ent in doc.ents:
     if ent.label_ == "NORP":
        original_ents.remove(ent)
   doc.ents = original_ents
   return(doc)
```

```python
# to add changed pipe in nlp.

nlp.add_pipe("remove_norp")
```

```python
# to see removed entity in end of the pipeline.

nlp.analyze_pipes()
```

```python
# NORP is remove now.

doc = nlp("Britain is a place. Mary is a doctor. Mary is british.")

for ent in doc.ents:
   print(f"Entity: {ent.text} | Label: {ent.label_}")
```
