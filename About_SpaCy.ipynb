```python
#1. Custom Components में क्या-क्या सीखना आवश्यक है (Detailed Learning Roadmap)
#Component Registration (@Language.component vs @Language.factory):
#Stateless Components (@Language.component): जब आपको बिना किसी बाहरी मॉडल के सिर्फ़ कोड या फ़िल्टर लगाना हो (जैसे एंटिटीज को हटाना या नियम लागू करना)।
#Stateful Components (@Language.factory): जब आपके कंपोनेंट को कोई बाहरी फ़ाइल, डेटाबेस या कॉन्फ़िगरेशन लोड करने की आवश्यकता हो।
#Pipeline Placement Control (nlp.add_pipe Parameters):
#यह समझना कि आपका कस्टम कंपोनेंट पाइपलाइन में किस स्थान पर चलेगा: before="ner", after="ner", first=True, या last=True।
#उदाहरण: अगर आपको NER से पहले टेक्स्ट को सुधारना है तो before="ner" यूज़ होगा; अगर NER के आउटपुट को फ़िल्टर करना है तो after="ner" यूज़ होगा।
#Extension Attributes (Doc, Token, and Span Extensions):
#Doc.set_extension(), Token.set_extension(), और Span.set_extension() का उपयोग करके spaCy के ऑब्जेक्ट्स में अपना नया डेटा/मेटाडेटा जोड़ना।
#3 मुख्य प्रकार सीखें:
#Attribute Extensions (Default values)
#Property Extensions (getter और setter फ़ंक्शन्स)
#Method Extensions (कस्टम फ़ंक्शन जो किसी ऑब्जेक्ट पर कॉल हो सके)
#ऐतिहासिक टेक्स्ट में उपयोग: किसी राजा या स्थान के टोकन पर कस्टम फ्लैग लगाना (जैसे: token._.is_historical_title = True)।
#Overlapping Entities & Entity Resolution (spacy.util.filter_spans):
#जब स्पैन ओवरलैप होते हैं (जैसे "पानीपत" और "पानीपत का युद्ध"), तो सही स्पैन चुनना और doc.ents को क्रैश किए बिना अपडेट करना।
#Custom Entity Normalization & Canonical Linking:
#अलग-अलग वर्तनी वाले ऐतिहासिक नामों (जैसे Acbar, Akbar, Akbar Shah) को पहचानकर उन्हें एक ही आईडी असाइन करने का लॉजिक कस्टम कंपोनेंट में लिखना।
#Serialization (to_disk / from_disk):
#अपने कस्टम कंपोनेंट को मॉडल के साथ हार्ड ड्राइव पर सेव करना ताकि nlp.to_disk() से पूरा कस्टम मॉडल री-यूज़ किया जा सके।
```
