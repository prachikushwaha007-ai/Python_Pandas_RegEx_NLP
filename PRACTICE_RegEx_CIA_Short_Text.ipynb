```python
import re
```

```python
# To Open File Using encoding.

with open('/content/drive/MyDrive/1.CIA_SHORT.txt', 'r', encoding = 'utf-8') as f:
     raw_text = f.read()
print(raw_text)
```

```python
# To Remove Extra Lines.

cleaned_text = re.sub(r'\n\s+\n', '\n\n', raw_text)
```

```python
# Tp Remove Extra Spaces and Tab Character.

cleaned_text = re.sub(r'[ \t]+', ' ', cleaned_text)
```

```python
# To Clean 'page' scatterd randomly.

cleaned_text = re.sub(r'page\s+\d', '', cleaned_text, re.I)
```

```python
# To Remove Header and Footer Pattern.

header_pattern = re.compile(r'(?i)declassified\s+in\s+part.*?(?=\n\n|\n[A-Z]|$)', re.I)

cleaned_text = re.sub(header_pattern, '', cleaned_text)
```

```python
# To Clean Randomly Scattered 'e'.

cleaned_text = re.sub(r'\be\b', '', cleaned_text)
```

```python
# To Remove Garbage Symbols.

garbage_symbols = re.compile(r'[^\w\s.,?!\'"\-_]')
cleaned_text = re.sub(garbage_symbols, '', cleaned_text)
```

```python
# To Remove Multiline Garbage.

multiline_garbage = re.compile(r'^\s*[a-zA-Z0-9]{1,2}\s*\n', re.MULTILINE)
cleaned_text = re.sub(multiline_garbage, '', cleaned_text)
```

```python
# To Clean Garbage Text Using lookahead and lazy match(*?).

garbage_text = re.compile(r'(?<=1962)[\s\S]*?(?=Chapter)')
cleaned_text = re.sub(garbage_text, ' ', cleaned_text)
```

```python
# To Clean Vowel Garbage.

vowel_garbage = re.compile(r'\b(?:[e]{2,}|[a]{2,}|[o]{2,}\b|\b(aoe)|eea|iea|hth|ae\b)', re.M)
cleaned_text = re.sub(vowel_garbage, '', cleaned_text)
```

```python
# To Clean Punctuation Garbage.

punctuation_garbage = re.compile(r'^\s*[.,\-_]?\s*[a-zA-Z0-9]{1,3}[,.]?\s*$', re.M)
cleaned_text = re.sub(punctuation_garbage, '', cleaned_text)
```

```python
# To Remove short Character Patterns/Garbage Words.

text = re.compile(r'[ \t]{2,}\b[a-zA-Z0-9]{1,2}\b(?=\s*$)', re.M)
cleaned_text = re.sub(text, '', cleaned_text)
```

```python
# To Remove Specific Standalone Words.

cleaned_text = re.sub(r'\b(a|as)\b', '', cleaned_text)
```

```python
# To Give an Output Path for Cleaned Text.

output_path = '/content/drive/MyDrive/1.CIA_Cleaned_SHORT.txt'
```

```python
# To Write Cleaned Text.

with open(output_path, 'w', encoding = 'utf-8') as f:
    f.write(cleaned_text)
print(cleaned_text)
```
