```python
# For generating interactive maps and geocoding for processing the longitude and lattitude.
# folium = library that makes HTML Maps.
# Nominatim class of geocoders module from geopy library use open street map data to find coordinates.
# Carolingian = unique user_agent name by developer.
# Nominatim prohibits request more than 1 sec., to takle this use 'time.sleep(1)' [import time].
# Use (if place:) before print places to handle the attribute error while none condition.
# Apply from geopy.extra.rate_limiter  import RateLimiter.
# For deciding Zoom level use fit_bounds()
```

```python
import time
import folium
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
```

```python
geolocator = Nominatim(user_agent='carolingean')
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)
```

```python
places = ['Tours', 'Auxerre', 'Fulda', 'Aachen', 'York']
```

```python
m = folium.Map(location = [47.3941, 0.6848], zoom_start = 5)

for i in places:
     if place:
        place = geolocator.geocode(i)
        print(place.latitude, place.longitude)
        tooltip = f'{i}'
        folium.Marker([place.latitude, place.longitude], popup=f'<strong>{i}</strong>', tooltip=tooltip, icon=folium.Icon(color='blue')).add_to(m)
```

```python
m.save('map.html')
```
