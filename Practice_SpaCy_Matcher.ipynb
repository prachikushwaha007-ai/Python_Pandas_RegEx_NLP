```python
import spacy
from spacy.matcher import Matcher
```

```python
!python -m spacy download en_core_web_lg
nlp = spacy.load("en_core_web_lg")
```

```python
# Matcher has vocab not doc.ents.

matcher = Matcher(nlp.vocab)
pattern = [{"LIKE_EMAIL": True}]
matcher.add("EMAIL_ADDRESS", [pattern])
```

```python
doc = nlp("This is an email address: wmattingly@aol.com")
matches = matcher(doc)
```

```python
print(matches)  # index=0 is lexeme a unique no.,then start and end token.
```

```python
# the label of EMAIL_ADDRESS gave that added into nlp vocab with this unique lexeme.
print(nlp.vocab[matches[0][0]].text)
```

```python
with open('/content/drive/MyDrive/Wiki_Data.txt', 'r') as f:
  text = f.read()
print(text)
```

```python
text = text.replace("\n\n", " ")
print(text)
```

```python
# to find the matches in doc vocab.

matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN"}]
matcher.add("PROPER_NOUN", [pattern])
doc = nlp(text)
matches = matcher(doc)
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
# to get multi word token add operator + and greedy.

matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN", "OP": "+"}]
matcher.add("PROPER_NOUN", [pattern], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
# add sort the matches key lambda by start token.

matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN", "OP": "+"}]
matcher.add("PROPER_NOUN", [pattern], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
# add verb pattern after proper noun.

matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN", "OP": "+"}, {"POS": "VERB"}]
matcher.add("PROPER_NOUNS", [pattern], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
# orth uses quote symbols pattern.

quote_symbols = ['"', '“', '”', "'", '‘', '’']

matcher = Matcher(nlp.vocab)

pattern1 = [{"ORTH": {"IN": quote_symbols}},
            {"ORTH": {"NOT_IN": quote_symbols}, "OP": "+"},
            {"ORTH": {"IN": quote_symbols}}
]

matcher.add("PROPER_NOUNS", [pattern1], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
# NORM have any symbol of quotes.

matcher = Matcher(nlp.vocab)

pattern = [{"NORM": {"IN": ['"']}},
            {"NORM": {"NOT_IN": ['"']}, "OP": "+"},
            {"NORM": {"IN": ['"']}}
]

matcher.add("QUOTE_PATTERN", [pattern], greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
with open("/content/drive/MyDrive/alice.txt", "r") as f:
  text2 = f.read()
print(text2)
```

```python
text2 = text2.replace("\n", " ")
print(text2)
```

```python
# capture statement within quotes.

matcher = Matcher(nlp.vocab)
pattern = [{"ORTH": "“"},
           {"IS_ALPHA": True, "OP": "+"},
           {"IS_PUNCT": True, "OP": "*"},
           {"ORTH": "”"}
]
matcher.add("PROPER_NOUNS", [pattern], greedy="LONGEST")
doc = nlp(text2)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches[:10]:
   print(match, doc[match[1]:match[2]])
```

```python
# capture statement within quotes.

speak_lemmas = ["think","say"]
matcher = Matcher(nlp.vocab)
pattern = [{"ORTH": "“"},
           {"IS_ALPHA": True, "OP": "+"},
           {"IS_PUNCT": True, "OP": "*"},
           {"ORTH": "”"},
           {"POS": "VERB", "LEMMA": {"IN": speak_lemmas}},
           {"POS": "PROPN", "OP": "+"},
           {"ORTH": "“"},
           {"IS_ALPHA": True, "OP": "+"},
           {"IS_PUNCT": True, "OP": "*"},
           {"ORTH": "”"}
          ]
matcher.add("PROPER_NOUNS", [pattern], greedy="LONGEST")
doc = nlp(text2)
matches = matcher(doc)
matches.sort(key = lambda x: x[1])
print(len(matches))
for match in matches:
   print(match, doc[match[1]:match[2]])
```
