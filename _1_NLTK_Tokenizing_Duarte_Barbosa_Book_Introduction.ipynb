```python
import nltk
```

```python
# Tokenization is a form of grouping.
# Tokenization can be done as word tokenizers & sentence tokenizers.
# lexicon amd corporas--
# corpora- body of text. eg; medical journals.
# lexicon- is like a dictionary; words and means.
```

```python
# import word and sent Tokenizers.

from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
# Secure code structure using try and except method from being crashed.

try:
    tokens = word_tokenize("This is a test sentence.")
except LookupError:
    nltk.download('punkt_tab')
    tokens = word_tokenize("This is a test sentence.")
```

```python
# file path of Duarte Barbosa's book Introduction

text = '/content/drive/MyDrive/INTRODUCTION.txt.txt'
```

```python
# Open Text File.

with open('/content/drive/MyDrive/INTRODUCTION.txt') as f :
     text = f.read()

print(text)
```

```python
# Tokenize by Sentences using for loop.

print(sent_tokenize(text))

for i in sent_tokenize(text):
    print(i)

len(sent_tokenize(text))
```

```python
# Tokenize by Word using for loop.

print(word_tokenize(text))

for i in word_tokenize(text):
    print(i)

len(word_tokenize(text))
```

```python
# import stopwords.

from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
# stopwords is a set of english stopwords that have already been predefined by NLTK.

stopwords = set(stopwords.words('english'))
print(stopwords)

len(stopwords)
```

```python
words = word_tokenize(text)
```

```python
# To filter the text: method one.

filtered_text = []

for w in words:
    if w not in stopwords:
       filtered_text.append(w)

print(filtered_text)

len(filtered_text)
```

```python
# To filter the text: method two.

filtered_text = [w for w in words if not w in stopwords]

print(filtered_text)

len(filtered_text)
```
