```python
import spacy
```

```python
with open ("/content/drive/MyDrive/Text_Barbosa.txt", "r", encoding="utf-8") as f:
    text = f.read()
print(text)
```

```python
text = text.replace("\n", " ")
print(text)
```

```python
!python -m spacy download en_core_web_md
nlp = spacy.load("en_core_web_md")
```

```python
doc = nlp(text)
```

```python
for ent in doc.ents:
   print(f"Entity: {ent.text} | Label: {ent.label_}")
print(len(doc.ents))
```

```python
import re
```

```python
# To match year occur in text

pattern = re.compile(r"\d{4}")
matches = re.finditer(pattern, text)
for match in matches:
    print(match)
```

```python
# To match specific words and further

pattern = re.compile(r"\bmoors\b(\s+\w+){5}", re.I)
matches = re.finditer(pattern, text)
for match in matches:
    print(match)
```

```python
# To find line starting from moors upto .

pattern = re.compile(r"\bmoors\b[^\.]*\.", re.I)
matches = re.finditer(pattern, text)
for match in matches:
    print(match)
```

```python
pattern = re.compile(r"^.*?\bmoors\b.*$", re.I)
matches = re.findall(pattern, text)
for match in matches:
    print(match)
```
