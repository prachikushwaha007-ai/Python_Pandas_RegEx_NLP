```python
import re
```

```python
text = "Prachi Kushwaha is learning technical skills. Rishi Kushwaha is going to buy laptop for her sister."
```

```python
pattern = r"(Prachi|Rishi) \w+"
```

```python
matches = re.finditer(pattern, text)
for match in matches:
   print(match)
```

```python
import spacy
from spacy.tokens import Span
nlp = spacy.blank("en")
doc = nlp(text)
original_ents = list(doc.ents)

mwt_ents = []
for match in re.finditer(pattern, doc.text):
    start, end = match.span()

    span = doc.char_span(start, end)
    if span is not None:
       mwt_ents.append((span.start, span.end, span.text))
for ent in mwt_ents:
    start, end, name = ent
    per_ent = Span(doc, start, end, label="PERSON")
    original_ents.append(per_ent)
    doc.ents = original_ents
for ent in doc.ents:
   print(f"Entity: {ent.text} | Label: {ent.label_}")
```

```python
from spacy.language import Language
from spacy.util import filter_spans
@Language.component("family_ner")
def family_ner(doc):
    original_ents = list(doc.ents)
    mwt_ents = []
    for match in re.finditer(pattern, doc.text):
        start, end = match.span()

        span = doc.char_span(start, end)
        if span is not None:
           mwt_ents.append((span.start, span.end, span.text))
    for ent in mwt_ents:
        start, end, name = ent
        per_ent = Span(doc, start, end, label="PERSON")
        original_ents.append(per_ent)
    filtered = filter_spans(original_ents)
    doc.ents = filter_spans(original_ents)
    return(doc)
```

```python
nlp = spacy.load("en_core_web_sm")
nlp.add_pipe("family_ner")
```

```python
doc2 = nlp(text)
for ent in doc2.ents:
   print(f"Entity: {ent.text} | Label: {ent.label_}")
```
