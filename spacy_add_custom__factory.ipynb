```python
# factory function that does somthing to doc object typically
```

```python
import spacy
from spacy.language import Language
```

```python
# add decorator with naming a custom component
@Language.component("cap_maker")
# create function that takes doc object
def cap_maker(doc):
    print(doc.text.capitalize())
    return (doc)

Language.component("cap_maker", func=cap_maker)
```

```python
nlp = spacy.blank("en")
```

```python
nlp.add_pipe("cap_maker")
```

```python
print(nlp.pipe_names)
```

```python
text = "this is a string"
```

```python
doc = nlp(text)
```

```python
# to save the model
nlp.to_disk("test_model")
```
