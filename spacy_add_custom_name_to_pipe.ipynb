```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
print(nlp.pipe_names)
```

```python
# add pipe of same naming gives an error
nlp.add_pipe("ner")
```

```python
# but in complex and robust model you'll want
# to have different ner models that do diff. things
# and they function at different positions in pipeline
# now to add another ner that has diff. name

nlp.add_pipe("ner", name="economic_ner")
```

```python
print(nlp.pipe_names)
```

```python
# to add pipe before lemmatizer
nlp.add_pipe("ner", name="eco_ner", before="lemmatizer")
```

```python
print(nlp.pipe_names)
```

```python
# to add pipe after parser
nlp.add_pipe("ner", name="tech_ner", after="parser")
```

```python
print(nlp.pipe_names)
```
